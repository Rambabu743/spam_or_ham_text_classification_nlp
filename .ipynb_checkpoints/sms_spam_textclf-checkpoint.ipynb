```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data=pd.read_csv('smsspamcollection.tsv', sep='\t')
data
```

```python
# Check target distribution

print(data['label'].value_counts())

sns.countplot(x='label', data=data)
plt.title("Target Distribution")
plt.show()
```

```python
data["message"]=data["message"].str.lower()
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem  import WordNetLemmatizer
```

```python
import re
lematizer=WordNetLemmatizer()
messages=[]
stop_words=stopwords.words("english")
for msg in data["message"]:
    msg_words=word_tokenize(re.sub("[^a-zA-Z]"," ",msg))
    msg_words=[lematizer.lemmatize(word) for word in msg_words if word not in stop_words]
    messages.append(" ".join(msg_words))
```

```python
len(messages)
```

```python
data['label'] = data['label'].map({
    'spam': 1,
    'ham': 0
})
```

```python
y=data["label"]
```

```python
data["message"]=messages
```

```python
X=data["message"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,stratify=y)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score
```

```python
models = {
    'LogisticRegression': Pipeline([
        ('tfidf', TfidfVectorizer(
            max_features=20000,
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95
        )),
        ('clf', LogisticRegression(
            class_weight='balanced',
            max_iter=1000
        ))
    ]),
    'LinearSVC': Pipeline([
        ('tfidf', TfidfVectorizer(
            max_features=20000,
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95
        )),
        ('clf', LinearSVC(
            class_weight='balanced'
        ))
    ])
}
```

```python

# Train and evaluate models

results = []

for name, model in models.items():

    print("="*60)
    print(f"MODEL: {name}")
    print("="*60)

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    print("Accuracy:", accuracy)

   

    results.append({
        'Model': name,
        'Accuracy': accuracy
    })
```

```python
import joblib
```

```python
joblib.dump(models['LogisticRegression'], 'spam_pipeline.pkl')
print("Model saved successfully")
```
